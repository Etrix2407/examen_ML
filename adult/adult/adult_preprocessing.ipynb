## 1. Importation des librairies <a id='section-1'></a>
`pandas` sert uniquement à charger le fichier texte source et à sauvegarder les CSV finaux. Toutes
les transformations utilisent `scikit-learn` : `ColumnTransformer`, `Pipeline`, `SimpleImputer`,
`StandardScaler`, `OneHotEncoder` et `LabelEncoder`.

```python
import pandas as pd

from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder, LabelEncoder
```

---
## 2. Chargement des données train/test <a id='section-2'></a>
`adult.data` (train) et `adult.test` (test) sont chargés depuis le répertoire courant (chemins
relatifs). Le caractère `?` est directement interprété comme valeur manquante (`NaN`) au chargement,
de façon à être géré ensuite par `SimpleImputer`. `adult.test` contient une ligne d'en-tête parasite
et des labels terminés par un point : on les harmonise au chargement, avant de laisser sklearn gérer
la suite du nettoyage.

```python
column_names = [
    'age', 'workclass', 'fnlwgt', 'education', 'education_num',
    'marital_status', 'occupation', 'relationship', 'race', 'sex',
    'capital_gain', 'capital_loss', 'hours_per_week', 'native_country', 'income'
]

def load_adult_dataset(path, skiprows=0):
    """Charge un fichier adult.data / adult.test brut (lecture pandas uniquement)."""
    df = pd.read_csv(path, header=None, names=column_names, sep=',',
                      skipinitialspace=True, na_values='?', skiprows=skiprows)
    df['income'] = df['income'].str.rstrip('.')
    return df

df_train = load_adult_dataset('adult.data')
df_test = load_adult_dataset('adult.test', skiprows=1)
```

`df_train` et `df_test` contiennent les 15 colonnes brutes, valeurs manquantes comprises (`NaN`).
Aucune ligne ni colonne n'est retirée manuellement à ce stade : cela sera pris en charge par le
pipeline `sklearn` défini ci-après.

---
## 3. Définition du pipeline de transformation (sklearn) <a id='section-3'></a>
Un `ColumnTransformer` sélectionne explicitement les colonnes numériques et catégorielles à
conserver ; `fnlwgt` et `income` ne figurant dans aucune des deux listes, elles sont automatiquement
exclues (paramètre `remainder='drop'` par défaut), sans appel à `DataFrame.drop`. Les valeurs
manquantes sont gérées par `SimpleImputer` (médiane pour le numérique, mode pour le catégoriel),
remplaçant toute suppression manuelle de lignes (`dropna`).

```python
numeric_features = ['age', 'education_num', 'capital_gain', 'capital_loss', 'hours_per_week']
categorical_features = ['workclass', 'education', 'marital_status', 'occupation',
                         'relationship', 'race', 'sex', 'native_country']

numeric_pipeline = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler()),
])

categorical_pipeline = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='ignore', sparse_output=False)),
])

preprocessor = ColumnTransformer(transformers=[
    ('num', numeric_pipeline, numeric_features),
    ('cat', categorical_pipeline, categorical_features),
], remainder='drop')
preprocessor
```

Le `ColumnTransformer` encapsule à la fois la **sélection** des variables pertinentes (exclusion
implicite de `fnlwgt`), l'**imputation** et la **transformation** (standardisation / One-Hot), en un
seul objet `sklearn`.

---
## 4. Encodage de la cible (sklearn) <a id='section-4'></a>
La cible `income` est encodée avec `LabelEncoder` (plutôt qu'une comparaison booléenne `pandas`),
ajusté sur le train puis appliqué au test.

```python
target_encoder = LabelEncoder()
target_encoder.fit(df_train['income'])
print("Classes :", target_encoder.classes_)

y_train = pd.Series(target_encoder.transform(df_train['income']), name='income')
y_test = pd.Series(target_encoder.transform(df_test['income']), name='income')
```

`LabelEncoder` associe `0` à `<=50K` et `1` à `>50K` (ordre alphabétique de ses classes), ce qui
correspond bien à la convention "revenu élevé = classe positive" recherchée.

---
## 5. Application du pipeline et sauvegarde <a id='section-5'></a>
On ajuste le `ColumnTransformer` sur `df_train` (l'intégralité du DataFrame, `fnlwgt` et `income`
étant ignorées automatiquement), on transforme `df_train` et `df_test`, puis on sauvegarde les 4
fichiers CSV attendus.

```python
preprocessor.fit(df_train)

feature_names = preprocessor.get_feature_names_out()

X_train = pd.DataFrame(preprocessor.transform(df_train), columns=feature_names)
X_test = pd.DataFrame(preprocessor.transform(df_test), columns=feature_names)

X_train.to_csv('X_train_preprocessed.csv', index=False)
X_test.to_csv('X_test_preprocessed.csv', index=False)
y_train.to_csv('y_train_preprocessed.csv', index=False)
y_test.to_csv('y_test_preprocessed.csv', index=False)

print("X_train :", X_train.shape)
print("X_test  :", X_test.shape)
print("4 fichiers CSV sauvegardés : X_train_preprocessed.csv, X_test_preprocessed.csv, "
      "y_train_preprocessed.csv, y_test_preprocessed.csv")
```

Les 4 fichiers CSV sont sauvegardés dans le répertoire courant. Aucune ligne n'a été supprimée
(toutes les valeurs manquantes ont été imputées par le pipeline `sklearn`) : `X_train` conserve les
32561 individus d'origine, `X_test` les 16281 individus d'origine. Ces fichiers serviront de point
de départ à la Phase 3 (sélection de variables).
